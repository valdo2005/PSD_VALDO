# BUSINESS UNDERSTANDING — ECGFiveDays
### 1. Ringkasan proyek
Proyek ini bertujuan membangun pipeline klasifikasi time-series dari dataset ECGFiveDays (UCR/UEA) untuk membedakan dua pola sinyal ECG. Fokusnya adalah demonstrasi teknik pengolahan sinyal, eksperimen model (baseline & deep learning), evaluasi kinerja, dan penyajian hasil secara interaktif melalui aplikasi Streamlit. Pekerjaan dilakukan di Jupyter Notebook untuk reproducibility.

### 2. Tujuan bisnis
Tujuan utama (non-klinikal): Menyediakan proof-of-concept otomasi klasifikasi pola ECG yang menonjolkan proses EDA, preprocessing, modeling (termasuk deep learning), evaluasi, dan deployment.

Tujuan teknis: Membangun model klasifikasi yang stabil dan dapat direplikasi yang:
- Memanfaatkan baseline klasik sebagai pembanding
- Menyertakan model deep learning sebagai model utama/penunjang.
- Menyediakan pipeline preprocessing sinyal yang dapat diuji dan dipilih di Streamlit.

Tujuan edukatif: Menunjukkan tiap langkah CRISP-DM secara jelas sehingga dapat dipakai sebagai materi pembelajaran atau demonstrasi tugas kuliah.

### 3. Lingkup & batasan
Lingkup: Dataset ECGFiveDays saja. Analisis difokuskan pada klasifikasi dua kelas yang tersedia pada dataset ini.

Batasan penting:

- Dataset bukan dataset klinis lengkap → proyek bukan untuk diagnosis medis.
- Data jumlahnya relatif kecil; deep learning perlu regularisasi / augmentasi agar tidak overfit.
- Waktu komputasi terbatas pada lingkungan lokal.

### 6. Kriteria sukses 
#### Teknis

- Tersedianya pipeline end-to-end (EDA → preprocessing → modeling → evaluasi → deployment).

- Baseline dikembangkan dan hasilnya terdokumentasi.

- Minimal satu model deep learning dilatih dan dievaluasi.

- Evaluasi menggunakan metrik: Accuracy, Precision, Recall, F1-score, Confusion Matrix; disertai cross-validation atau repeated experiments.

- Target semu: baseline ≥ 80%; model DL diperkirakan dapat menyamai/meningkatkan performa jika dioptimalkan. (Catatan: angka target harus disesuaikan setelah EDA.)

#### Non-teknis

- Aplikasi Streamlit yang memungkinkan: upload file sinyal, preview sinyal, pilih preprocessing/pilih model, prediksi kelas + probabilitas, dan unduh hasil.

### 7. Risiko utama & mitigasi
- Overfitting model deep learning

Mitigasi: regularization, early stopping, dropout, data augmentation, cross-validation, dan monitoring learning curves.

- Dataset terlalu kecil untuk deep learning

Mitigasi: gunakan transfer learning jika ada model pretrained (jarang untuk 1D ECG); atau prioritaskan arsitektur ringan + augmentasi.

- Noisy ECG / artefak

Mitigasi: filtering (bandpass), smoothing, outlier removal, normalization.

- Kesalahan interpretasi sebagai hasil klinis

Mitigasi: eksplisit menyatakan batasan dalam laporan dan UI.

- Perbedaan sampling rate / panjang sinyal

Mitigasi: resampling / padding / truncation yang konsisten.

### 8. Output Streamlit — fitur yang direncanakan 
Fitur utama di app:

- Upload file sinyal (.txt/.csv/.npy) atau pilih contoh dataset.

- Preview sinyal (plot) dan metadata (panjang, sampling rate asumsi).

- Pilihan preprocessing (z-norm, bandpass, smoothing, resample).

- Pilih model (Baseline DTW / Feature-based / 1D-CNN).

- Tombol Predict → menampilkan kelas prediksi, confidence/probability, dan plot overlay sinyal contoh dari kelas terdekat (untuk explainability).

- Tampilkan metrics singkat (akurasi test set) dan confusion matrix dari model terpilih (precomputed).

- Download hasil prediksi (CSV).
